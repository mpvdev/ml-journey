# Model Representation: Pokémon Example

This companion lab applies the one-variable linear regression ideas from `Model Representation.ipynb` to Pokémon data. We use `Attack` to predict the `Total` score.

## Training data

Each Pokémon is one training example. The feature $x^{(i)}$ is its Attack value, and the target $y^{(i)}$ is its Total score.

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

plt.style.use("./deeplearning.mplstyle")
pokemon = pd.read_csv("../../../../data/pokemon.csv")
pokemon_train = pokemon[["Name", "Attack", "Total"]].dropna()

x_train = pokemon_train["Attack"].to_numpy(dtype=float)
y_train = pokemon_train["Total"].to_numpy(dtype=float)
m = x_train.shape[0]

print(pokemon_train.head())
print(f"Number of training examples: {m}")
```

```python
plt.scatter(x_train, y_train, marker=".", alpha=0.6)
plt.title("Pokémon Attack and Total Score")
plt.xlabel("Attack")
plt.ylabel("Total score")
plt.show()
```

## Model function

The linear model is $f_{w,b}(x) = wx + b$. Different values of $w$ and $b$ produce different prediction lines.

```python
def compute_model_output(x, w, b):
    m = x.shape[0]
    f_wb = np.zeros(m)

    for i in range(m):
        f_wb[i] = w * x[i] + b

    return f_wb

w = 2.0
b = 300.0
predictions = compute_model_output(x_train, w, b)
order = np.argsort(x_train)
```

```python
plt.scatter(x_train, y_train, marker=".", alpha=0.5, label="Actual Pokémon")
plt.plot(x_train[order], predictions[order], c="b", label="Linear model")
plt.title("Pokémon Linear Model")
plt.xlabel("Attack")
plt.ylabel("Total score")
plt.legend()
plt.show()
```

## Prediction

Use the model to estimate the Total score for a Pokémon with an Attack value of 100.

```python
attack_value = 100
predicted_total = w * attack_value + b
print(f"Predicted Total score: {predicted_total:.0f}")
```

## Conclusion

The feature-target relationship is the same as in the original lab: a linear regression model maps one numeric input to a prediction using parameters $w$ and $b$.
